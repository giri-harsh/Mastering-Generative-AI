## Semantic similarity calculator using Sentence Transformers


```python
from sentence_transformers import sentence_transformer,util
from dotenv import load_dotenv
from langchain_huggingface import HuggingFaceEndpointEmbeddings
from sklearn.metrics.pairwise import cosine_similarity 
```

```python
load_dotenv()
embeddings = HuggingFaceEndpointEmbeddings(
    model= "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
def compare(t1,t2):
    embeddings1 = embeddings.embed_query(t1)
    embeddings2 = embeddings.embed_query(t2)
    similarity_score = cosine_similarity([embeddings1],[embeddings2])[0][0]
    return similarity_score
```

```python
sentence1 = "I have a dog"
sentence2 = "I have a cat"
```

```python
score = compare(sentence1,sentence2)
print("Similary Score : ",score)


if score > 0.90:
    print("Result: Almost Identical")

elif score > 0.75:
    print("Result: Very Similar")

elif score > 0.50:
    print("Result: Related")

else:
    print("Result: Different")
```
