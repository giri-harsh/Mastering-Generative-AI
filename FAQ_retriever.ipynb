# FAQ Retriever


```python
from dotenv import load_dotenv
from langchain_huggingface import HuggingFaceEmbeddings
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import sentence_transformer
```

```python
faqs = [
    {
        "question": "How do I reset my account password?",
        "answer": "Click on 'Forgot Password' on the login page and follow the instructions."
    },

    {
        "question": "How can I contact customer support?",
        "answer": "You can contact customer support through email or live chat."
    },

    {
        "question": "How do I update my profile information?",
        "answer": "Go to Account Settings and edit your profile details."
    },

    {
        "question": "How can I cancel my subscription?",
        "answer": "Open Billing Settings and select Cancel Subscription."
    },

    {
        "question": "What payment methods are accepted?",
        "answer": "We accept credit cards, debit cards, and PayPal."
    },

    {
        "question": "How do I change my email address?",
        "answer": "Navigate to Account Settings and update your email."
    },

    {
        "question": "How do I download my invoices?",
        "answer": "Invoices can be downloaded from the Billing section."
    },

    {
        "question": "Can I upgrade my plan later?",
        "answer": "Yes, plans can be upgraded anytime from Billing Settings."
    }
]
```

```python
load_dotenv()
```

```python
embed = HuggingFaceEmbeddings(
    model = "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
que = [faq["question"] for faq in faqs]
# print(que)
# type(que)
que_embeding = embed.embed_documents(que)
```

```python
querry = "How to track order"
```

```python
querry_embedding = embed.embed_query(querry)
```

```python
scores = []
for i,j in enumerate(que_embeding):
    sim_score = cosine_similarity([querry_embedding],[j])[0][0]
    scores.append((i,sim_score))
```

```python
scores.sort(key=lambda x: x[1], reverse=True)


# ------------------------------------------------------------
# Display Top 3 Matching FAQs
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("TOP 3 FAQ MATCHES")
print("=" * 60)

for rank, (index, score) in enumerate(scores[:3], start=1):

    print(f"\nRank #{rank}")
    print(f"Similarity Score: {score:.4f}")
    print(f"Question: {faqs[index]['question']}")
    print(f"Answer: {faqs[index]['answer']}")
    print("-" * 60)
```
