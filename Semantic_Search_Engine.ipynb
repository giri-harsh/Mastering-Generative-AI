## Semantic Search Engine

```python
from dotenv import load_dotenv
from langchain_huggingface import HuggingFaceEmbeddings
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import sentence_transformer
```

```python
load_dotenv()
```

```python
documents = [
    "Python is a popular programming language used for data science and web development.",

    "Machine learning enables computers to learn patterns from data without being explicitly programmed.",

    "Deep learning is a subset of machine learning that uses neural networks with many layers.",

    "Natural Language Processing allows computers to understand and generate human language.",

    "Basketball is a team sport where players score points by shooting a ball through a hoop.",

    "The Taj Mahal is one of the most famous monuments in India and a UNESCO World Heritage Site.",

    "Cloud computing provides on-demand access to computing resources over the internet.",

    "Git is a version control system used by developers to track changes in code.",

    "Data visualization helps communicate insights using charts, graphs, and dashboards.",

    "Artificial Intelligence aims to create systems capable of performing tasks that normally require human intelligence."
]



```

```python
embeddings = HuggingFaceEmbeddings(
    model = "sentence-transformers/all-MiniLM-L6-v2"
)

```

```python
document_embedding = []
for i in documents :
    embedding = embeddings.embed_query(i)
    document_embedding.append(embedding)
```

```python
query = "I love Basketball"
```

```python
query_embedding = embeddings.embed_query(query)
```

```python
scores = []

for i, embedding in enumerate(document_embedding):

    similarity_score = cosine_similarity(
        [query_embedding],
        [embedding]
    )[0][0]

    scores.append((i, similarity_score))
```

```python
scores.sort(key=lambda x: x[1], reverse=True)

for index, score in scores[:3]:
    print(f"Score: {score:.4f}")
    print(documents[index])
    print()
```
