# Quiz Generator using LangChain LCEL

## Objective

Build a multi-step AI workflow that takes a topic as input and automatically generates:

1. Detailed Notes
2. Summary
3. Quiz Questions

---

## Workflow

```text
User Topic
    ↓
Generate Notes
    ↓
Generate Summary
    ↓
Generate Quiz Questions
    ↓
Display Quiz
```

---



````python
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
````

````python
llm = HuggingFaceEndpoint(
    repo_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation",
    huggingfacehub_api_token="lol"
)

model = ChatHuggingFace(
    llm=llm
)
````

````python
# prompt1
temp1 = PromptTemplate(
    template="Give me 5 Line technical notes on the following {topic}",
    input_variables=['topic']
)
temp2 = PromptTemplate(
    template="Provide a short and easy to learn summary on {text}",
    input_variables=['text']
)
temp3 = PromptTemplate(
    template="give me 5 questions only based on {text} difficult level moderate",
    input_variables=['text']
)
````

````python
parser = StrOutputParser()
````

````python
chain = temp1 | model |parser | temp2 | model |parser | temp3 | model|parser
result = chain.invoke({'topic':"Machine Learning"})
print(result)
````
