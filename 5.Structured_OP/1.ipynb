```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from typing import TypedDict,Annotated
```

```python
load_dotenv()
model = ChatGroq(
    model="meta-llama/llama-4-scout-17b-16e-instruct"
)

```

```python
class Review(TypedDict):
    summary : str
    sentiment :str

```

```python
stuctured_model = model.with_structured_output(Review)
result = stuctured_model.invoke(
    "I recently bought the Moto G85 and have been using it for a few weeks. The display is bright and smooth, making videos and scrolling enjoyable. Battery life easily lasts a full day with regular use. Performance is good for everyday tasks like browsing, social media, and light gaming. The camera performs well in daylight, though low-light photos can be average. Overall, it offers excellent value for the price and feels like a solid mid-range smartphone. "
)
print(result)
```

```python
print(f"Summary {result['summary']}\n")

print(f"Semtiment : {result['sentiment']}\n")
```
